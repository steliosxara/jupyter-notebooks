## Το Πρόβλημα

Να κατασκευαστεί πρόγραμμα που υπολογίζει το παραγοντικό ενός αριθμού.

### Παράδειγμα

5! = 5 × 4 × 3 × 2 × 1 = 120

## Είσοδος - Επεξεργασία - Έξοδος (IPO)

```text
Είσοδος → Επεξεργασία → Έξοδος
   5     → 5×4×3×2×1 → 120
```

````python
number = int(input('Δώσε έναν αριθμό: '))
print('Έδωσες:', number)
````

## Δομή Επιλογής (if)

````python
number = -3

if number < 0:
    print('Μη αποδεκτός αριθμός')
else:
    print('Συνέχισε τον υπολογισμό')
````

## Δομή Επανάληψης (for)

````python
number = 5
factorial = 1

for i in range(1, number + 1):
    factorial = factorial*i

print(factorial)
````

### Πίνακας Ιχνών

| i | factorial |
|---|--:|
|1|1\*1=1|
|2|1\*2=2|
|3|2\*3=6|
|4|6\*4=24|
|5|24\*5=120|

## Συναρτήσεις
Συλλογή εντολών με όνομα που εκτελούν μια λειτουγία.

````python
def factorial(n):
    result = 1

    for i in range(1, n + 1):
        result *= i

    return result

print(factorial(5))
````

## Αναδρομή
Κλήση μιας συνάρτησης μέσα στο σώμα των εντολών της.

````python
def factorial_recursive(n):

    if n == 0 or n == 1:
        return 1

    return n * factorial_recursive(n-1)

print(factorial_recursive(5))
````

## Διαδραστική Άσκηση 4

Παρατηρήστε την αναδρομική κλήση.  
**Ερώτηση:**  
Ποια θα είναι η σειρά εμφάνισης των αριθμών;

````python
def countdown(n):

    if n == 0:
        print('Τέλος')
        return

    print(n)
    countdown(n-1)

countdown(3)
````
