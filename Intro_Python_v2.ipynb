## Δομή Επανάληψης (for)
Επανάληψη για συγκεκριμένο αριθμό φορών

```python
number = 5
factorial = 1

for i in range(1, number + 1):
    factorial *= i # συντομογραφία του factorial = factorial * i

print(factorial)
```

### Πίνακας Ιχνών

| i | factorial \*= i|
|---|:--:|
|1|1\*1=1|
|2|1\*2=2|
|3|2\*3=6|
|4|6\*4=24|
|5|24\*5=120|

## Συναρτήσεις
Συλλογή εντολών με όνομα που εκτελούν μια λειτουγία.

```python
def factorial(n):
    result = 1

    for i in range(1, n + 1):
        result *= i

    return result

print(factorial(5))
```

## Αναδρομή
Κλήση μιας συνάρτησης μέσα στο σώμα των εντολών της.

```python
def factorial_recursive(n):

    if n == 0 or n == 1:
        return 1

    return n * factorial_recursive(n-1)

print(factorial_recursive(5))
```

## Διαδραστική Άσκηση 4

Παρατηρήστε την αναδρομική κλήση.  
**Ερώτηση:**  
Ποια θα είναι η σειρά εμφάνισης των αριθμών;

```python
def countdown(n):

    if n == 0:
        print('Τέλος')
        return

    print(n)
    countdown(n-1)

countdown(3)
```

## Ολοκληρωμένο Πρόγραμμα

```python
def factorial(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result

while True:

    print('\n=== ΜΕΝΟΥ ===')
    print('1. Υπολογισμός Παραγοντικού')
    print('2. Έξοδος')

    choice = input('Επιλογή: ')

    if choice == '1':

        number = int(input('Δώσε αριθμό: '))

        if number < 0:
            print('Μη αποδεκτός αριθμός')
        else:
            print('Αποτέλεσμα:', factorial(number))

    elif choice == '2':
        print('Τερματισμός')
        break

    else:
        print('Λανθασμένη επιλογή')
```

## Εξηγήσεις με σχόλια

```python
# Συνάρτηση που υπολογίζει το παραγοντικό ενός αριθμού n
def factorial(n):
    result = 1  # Αρχικοποίηση του αποτελέσματος με 1

    # Πολλαπλασιάζουμε όλους τους αριθμούς από το 1 μέχρι το n
    for i in range(1, n + 1):
        result *= i

    # Επιστρέφουμε το τελικό αποτέλεσμα
    return result


# Επανάληψη του μενού μέχρι ο χρήστης να επιλέξει έξοδο
while True:

    # Εμφάνιση του μενού επιλογών
    print('\n=== ΜΕΝΟΥ ===')
    print('1. Υπολογισμός Παραγοντικού')
    print('2. Έξοδος')

    # Ανάγνωση της επιλογής του χρήστη
    choice = input('Επιλογή: ')

    # Επιλογή 1: Υπολογισμός παραγοντικού
    if choice == '1':

        # Εισαγωγή αριθμού από τον χρήστη
        number = int(input('Δώσε αριθμό: '))

        # Έλεγχος αν ο αριθμός είναι αρνητικός
        if number < 0:
            print('Μη αποδεκτός αριθμός')
        else:
            # Κλήση της συνάρτησης factorial και εμφάνιση του αποτελέσματος
            print('Αποτέλεσμα:', factorial(number))

    # Επιλογή 2: Τερματισμός προγράμματος
    elif choice == '2':
        print('Τερματισμός')
        break  # Έξοδος από την επανάληψη

    # Διαχείριση λανθασμένης επιλογής
    else:
        print('Λανθασμένη επιλογή')
```
